**Ejercicio 12: Calculadora modular**

Crear funciones:

• sumar

• restar

• multiplicar

• dividir


```python
def suma(a,b):
    if isinstance(a,(int,float)) and isinstance(b,(int,float)): #<-- Comprueba que los valores a y b son instancias de tipo int o float
        return a + b
    else:
        return 0
    
def resta(a,b):
    if isinstance(a,(int,float)) and isinstance(b,(int,float)): #<-- Comprueba que los valores a y b son instancias de tipo int o float
        return a - b
    else:
        return 0
    
def multiplicacion(a,b):
    if isinstance(a,(int,float)) and isinstance(b,(int,float)): #<-- Comprueba que los valores a y b son instancias de tipo int o float
        return a * b
    else:
        return 0

def division(a,b):
    if isinstance(a,(int,float)) and isinstance(b,(int,float)): #<-- Comprueba que los valores a y b son instancias de tipo int o float
        return a / b
    else:
        return 0
    
a = float(input("Dame un valor "))
op = input("Introduce una operación (+ , - , * , / ) ")
b = float(input("Dame otro valor "))
c = 0
match(op):
    case "+":
        c = suma(a,b)
    case "-":
        c = resta(a,b)
    case "*":
        c = multiplicacion(a,b)
    case "/":
        c = division(a,b)
    case _:
        print("Operación no válida")
        
print(f"{a} {op} {b} = {c}")
```

**Ejercicio 13: Es primo**

Crear:

es_primo(numero)

que devuelva verdadero o falso.


```python
def es_primo(x):
    if (x <= 1):
        return False
    else:
        for i in range(2, int(x ** 0.5)+1):
            if x % i == 0:
                return False
        return True

x = int(input("Dime un número "))
esPrimo = es_primo(x)

if(esPrimo):
    print("El número es primo")
else:
    print("El número no es primo")
```
