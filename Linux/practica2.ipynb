# Práctica 2: Gestión de archivos y permisos en Linux

Imaginemos que trabajamos en una pequeña empresa y necesitamos organizar la documentación de nuestros empleados.

Para ello, comenzaremos creando la siguiente estructura de directorios:

```text
empresa/
├── empleados/
├── administracion/
├── proyectos/
│   ├── activos/
│   └── archivados/
└── temporal/
```
Para ello, comenzaremos usando de nuevo el comando mkdir.

````python
mkdir -p empresa/empleados
mkdir -p empresa/administracion
mkdir -p empresa/proyectos/activos
mkdir -p empresa/proyectos/archivados
mkdir -p empresa/temporal
````

Para comprobar que hemos creado correctamente la estructura, ejecutaremos:

````python
find empresa -type d
````

Guardaremos una captura de pantalla del resultado.

2. Creación y escritura de archivos
Ahora vamos a crear algunos documentos de nuestra empresa.

Primero, entraremos en el directorio de empleados:

````python
cd empresa/empleados
````

El comando touch permite crear archivos vacíos. También permite actualizar la fecha de modificación de un archivo existente.

Ahora escribiremos información dentro del archivo:

````python
echo "Nombre: Juan" > empleado1.txt
echo "Departamento: Informatica" >> empleado1.txt
echo "Puesto: Tecnico" >> empleado1.txt
````

¡Cuidado! Hay una diferencia importante entre los operadores > y >>:
- ">" escribe en el archivo, sustituyendo todo su contenido anterior.
- ">>" añade información al final del archivo, sin borrar lo que ya había.

Para comprobar el resultado:

````python
cat empleado1.txt
````

Guardaremos una captura de pantalla.

3. Copiar, mover y renombrar archivos
Ahora necesitamos realizar algunas operaciones con nuestros documentos.

Primero, copiaremos el archivo del empleado al directorio de administración:

````python
cp empleado1.txt ../administracion/
````

El comando cp permite copiar archivos y directorios.

A continuación, volveremos al directorio principal de nuestra empresa:

````python
cd ..
````

Ahora moveremos el documento que hemos copiado a la carpeta de proyectos activos:

````python
mv administracion/empleado1.txt proyectos/activos/
````

El comando mv permite mover archivos y también cambiarles el nombre.

Vamos a cambiar el nombre del archivo:

````python
mv proyectos/activos/empleado1.txt proyectos/activos/ficha_juan.txt
````

Para comprobar el resultado:

````python
find . -type f
````

Guardaremos una captura de pantalla.

4. Buscar información dentro de archivos

Imaginemos que nuestra empresa tiene muchos empleados y necesitamos localizar rápidamente a aquellos que pertenecen a un departamento concreto.

Primero, crearemos un archivo con varios empleados:

````python
echo "Juan - Informatica" > empleados/listado.txt
echo "Maria - Administracion" >> empleados/listado.txt
echo "Pedro - Informatica" >> empleados/listado.txt
echo "Laura - Recursos Humanos" >> empleados/listado.txt
echo "Ana - Informatica" >> empleados/listado.txt
````

Ahora utilizaremos el comando grep:

````python
grep "Informatica" empleados/listado.txt
````

Este comando permite buscar una palabra o expresión dentro de un archivo.

Si queremos saber en qué línea aparece cada resultado, podemos utilizar:

````python
grep -n "Informatica" empleados/listado.txt
````

También podemos contar cuántas líneas tiene nuestro archivo:

````python
wc -l empleados/listado.txt
````

El argumento -l indica que queremos contar líneas.

Otros argumentos interesantes son:
- -w: cuenta palabras.
- -c: cuenta caracteres (bytes).

Guardaremos una captura de pantalla con el resultado de las búsquedas.

5. Consultar y modificar permisos
En Linux, los archivos y directorios tienen permisos que determinan quién puede leerlos, modificarlos o ejecutarlos.

Para consultar los permisos de nuestros archivos, ejecutaremos:

````python
ls -l empleados/listado.txt
````

Veremos algo parecido a esto:

````python
-rw-r--r-- 1 alumno alumno 120 Nov 8 12:00 listado.txt
````

Nos interesa especialmente la primera parte -rw-r--r--

Los permisos se organizan en tres grupos:

| Grupo | Significado |
|---|---|
| Propietario | Usuario dueño del archivo |
| Grupo | Usuarios pertenecientes al grupo |
| Otros | Resto de usuarios |

Los permisos disponibles son:

| Permiso | Significado |
|---|---|
| r | Lectura (read) |
| w | Escritura (write) |
| x | Ejecución (execute) |

Ahora vamos a modificar los permisos de nuestro archivo:

````python
chmod 600 empleados/listado.txt
````

El comando chmod permite modificar los permisos de archivos y directorios.

El número 600 significa:
- Primer número (6): el propietario puede leer y escribir.
- Segundo número (0): el grupo no tiene permisos.
- Tercer número (0): los demás usuarios no tienen permisos.

Volvemos a consultar los permisos:

````python
ls -l empleados/listado.txt
````

El resultado debería comenzar por -rw-------

Guardaremos una captura de pantalla.

6. Eliminación de archivos
Por último, vamos a aprender a eliminar archivos.

Crearemos un archivo temporal en la carpeta empresa/temporal, asumiendo que estamos en la carpeta empresa:

````python
echo "Este archivo es temporal" > ./temporal/temporal.txt
````

Comprobaremos su existencia con el comando ls y lo eliminaremos:

````python
rm temporal.txt
````

El comando rm permite eliminar archivos.

Para comprobar que ha desaparecido volveremos a ejecutar ls.

**Importante**: Linux no suele enviar los archivos eliminados mediante rm a una papelera de reciclaje. Por tanto, debemos tener cuidado al utilizarlo.
