# Crear un sistema de archivos

Lo primero que haremos será crear esta estructura de carpetas en nuestra terminal Linux:

```text
linux_practica/
├── documentos/
│   └── notas/
└──red/
```
Para hacerlo, ejecutaremos estos comandos:

````python
mkdir ./linux_practica
mkdir -p ./linux_practica/documentos/notas
mkdir -p ./linux_practica/red
````

Cuando escribimos ./ estamos diciendo "Desde la carpeta en la que estás ahora", por ejemplo, si estamos en la carpeta "casa", la ruta final para nosotros sería /casa/linux_practica, pero sigue pudiendo ser accedida como ./linux_practica siempre y cuando estemos en la carpeta casa. Para verificar el resultado ejecutaremos el siguiente comando, y guardaremos una captura de pantalla.

````python
find ./linux_practica -type d
````

Existen comandos que aceptan una serie de configuraciones a través de lo que llamamos argumentos. En este caso, el comando hace lo siguiente:

- find "busca"
- Primer argumento: ./linux_practica "en la carpeta linux_practica"
- Segundo argumento: -type "archivos del tipo"
- Tercer argumento: d "directorio -> carpeta"

En lenguaje natural le estamos diciendo, por tanto, a la máquina "Busca en la carpeta linux_practica archivos del tipo carpeta". El orden de los argumentos es extremadamente importante, de forma habitual no se puede cambiar.

Lo siguiente que haremos será borrar el contenido de la pantalla con el comando clear y entrar en el directorio que hemos creado y guardamos captura:

````python
clear
cd ./linux_practica
````

Lo siguente que haremos será imprimir texto en la pantalla y crearemos un archivo de texto que contenga ese texto. Por ejemplo:

````python
echo "Hola mundo"
echo "Hola mundo" > ./documentos/notas/archivo.txt
````

Lo que estamos haciendo arriba es, primero, que el texto que nosotros pongamos lo escriba la máquina con el comando echo, en el segundo hacemos lo mismo, pero vertimos (>) en un archivo de texto cuya ruta es ./documentos/notas/archivo.txt el texto que hemos escrito. Ahora, ejecutamos el siguiente comando para comprobar el resultado, luego guardamos captura de pantalla y por último ejecutamos de nuevo clear.

````python
cat ./documentos/notas/archivo.txt
````

Significado de los comandos:

- cd -> cdirectory -> See Directory -> Ver Directorio

- cat -> See At -> Ver En
